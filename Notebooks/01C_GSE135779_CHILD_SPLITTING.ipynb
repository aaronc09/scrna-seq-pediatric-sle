```python
import scanpy as sc
import sys
sys.path.insert(0, BASE_DIR + '/Notebooks')
from sample_lists import CHILD_HEALTHY_SAMPLES, CHILD_SLE_SAMPLES, ADULT_HEALTHY_SAMPLES, ADULT_SLE_SAMPLES

adata = sc.read_h5ad(
    f"{BASE_DIR}/child_individual_h5ad/adata_child_final_liana.h5ad"
)

healthy_samples = CHILD_HEALTHY_SAMPLES

sle_samples = CHILD_SLE_SAMPLES

adata.obs["condition"] = "unknown"
adata.obs.loc[adata.obs["sample"].isin(healthy_samples), "condition"] = "healthy"
adata.obs.loc[adata.obs["sample"].isin(sle_samples), "condition"] = "SLE"

print(adata.obs["condition"].value_counts())
print(adata.obs[["sample", "condition"]].drop_duplicates().sort_values("sample"))
```

```python
adata_child_healthy = adata[adata.obs["condition"] == "healthy"].copy()
adata_child_sle = adata[adata.obs["condition"] == "SLE"].copy()

print("Healthy:", adata_child_healthy)
print("SLE:", adata_child_sle)
```

```python
healthy_path = f"{BASE_DIR}/child_individual_h5ad/adata_child_healthy_liana.h5ad"
sle_path = f"{BASE_DIR}/child_individual_h5ad/adata_child_sle_liana.h5ad"

adata_child_healthy.write_h5ad(healthy_path)
adata_child_sle.write_h5ad(sle_path)

print("Saved healthy:", healthy_path)
print("Saved SLE:", sle_path)
```

```python
import scanpy as sc
import pandas as pd
from pathlib import Path

files = {
    "child_sle": Path(f"{BASE_DIR}/child_individual_h5ad/adata_child_sle_liana.h5ad"),
    "child_healthy": Path(f"{BASE_DIR}/child_individual_h5ad/adata_child_healthy_liana.h5ad"),
}

for name, path in files.items():
    print("=" * 80)
    print(name)
    print(path)

    adata = sc.read_h5ad(path)

    if "cell_type" not in adata.obs.columns:
        print("No 'cell_type' column found.")
        print("Available obs columns:")
        print(adata.obs.columns.tolist())
        continue

    cell_type_counts = (
        adata.obs["cell_type"]
        .value_counts(dropna=False)
        .reset_index()
    )

    cell_type_counts.columns = ["cell_type", "n_cells"]

    print("\nUnique cell types:")
    print(cell_type_counts["cell_type"].tolist())

    print("\nCell type counts:")
    display(cell_type_counts)

    print("\nTotal cells:", adata.n_obs)
```
